```python
import numpy as np

y = np.array([3, 5, 7])
y_pred = np.array([2, 5, 9])

error = y - y_pred

squared_error = error ** 2

mse = np.mean(squared_error)

print("error:", error)
print("squared error:", squared_error)
print("MSE:", mse)
```

```python
x = np.array([1, 2, 3])
y = np.array([3, 5, 7])

def mse(y, y_pred):
    return np.mean((y - y_pred) ** 2)

# 모델 1
w = 1
b = 0

y_pred = w * x + b

print("actualy:", y)
print("model 1:", y_pred)
print("MSE:", mse(y, y_pred))

# 모델 2
w = 2
b = 0

y_pred = w * x + b

print("model 2:", y_pred)
print("MSE:", mse(y, y_pred))

# 모델 3
w = 2
b = 1

y_pred = w * x + b

print("model 3:", y_pred)
print("MSE:", mse(y, y_pred))
```

```python
x = np.array([1, 2, 3])
y = np.array([3, 5, 7])

b = 1

for w in [0, 0.5, 1, 1.5, 2, 2.5, 3]:
    y_pred = w * x + b
    loss = np.mean((y - y_pred) ** 2)

    print(f"w={w:.1f}, loss={loss:.3f}, y_pred={y_pred}")
```
