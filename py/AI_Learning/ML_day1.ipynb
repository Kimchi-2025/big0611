```python
import numpy as np
import matplotlib.pyplot as plt

# 데이터
x = np.array([1., 2., 3., 4.])
y = np.array([3., 5., 7., 9.])

# 초기 parameter
w = 0.0
b = 0.0

# hyperparameter
learning_rate = 0.01
epochs = 1000

n = len(x)

# 기록용
w_history = []
b_history = []
loss_history = []

for epoch in range(epochs):

    # 1. prediction
    y_pred = w * x + b

    # 2. loss
    loss = np.mean((y_pred - y) ** 2)

    # 기록
    w_history.append(w)
    b_history.append(b)
    loss_history.append(loss)

    # 3. gradient
    dw = (2 / n) * np.sum((y_pred - y) * x)
    db = (2 / n) * np.sum(y_pred - y)

    # 4. parameter update
    w = w - learning_rate * dw
    b = b - learning_rate * db


# -----------------------
# 그래프
# -----------------------

plt.figure(figsize=(8, 5))

plt.plot(w_history, loss_history, color="blue", alpha=0.7)

# 각 epoch의 점
plt.scatter(
    w_history[::20],
    loss_history[::20],
    color="red",
    s=20
)

# 시작점
plt.scatter(
    w_history[0],
    loss_history[0],
    color="green",
    s=100,
    label="start"
)

# 마지막점
plt.scatter(
    w_history[-1],
    loss_history[-1],
    color="black",
    s=100,
    label="end"
)

plt.xlabel("w")
plt.ylabel("loss")
plt.title("Gradient Descent: w vs loss")
plt.legend()
plt.grid()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 데이터
x = np.array([1., 2., 3., 4.])
y = np.array([3., 5., 7., 9.])

# 초기 parameter
w = 0.0
b = 0.0

learning_rate = 0.01
epochs = 1000

n = len(x)

# 기록
w_history = []
b_history = []
loss_history = []

for epoch in range(epochs):

    # prediction
    y_pred = w * x + b

    # loss
    loss = np.mean((y_pred - y) ** 2)

    # 기록
    w_history.append(w)
    b_history.append(b)
    loss_history.append(loss)

    # gradient
    dw = (2 / n) * np.sum((y_pred - y) * x)
    db = (2 / n) * np.sum(y_pred - y)

    # update
    w = w - learning_rate * dw
    b = b - learning_rate * db


# -----------------------
# loss surface 계산
# -----------------------

w_values = np.linspace(-1, 4, 100)
b_values = np.linspace(-2, 4, 100)

W, B = np.meshgrid(w_values, b_values)

Loss = np.zeros_like(W)

for i in range(W.shape[0]):
    for j in range(W.shape[1]):
        y_pred = W[i, j] * x + B[i, j]
        Loss[i, j] = np.mean((y_pred - y) ** 2)


# -----------------------
# 그래프
# -----------------------

plt.figure(figsize=(8, 6))

# loss 등고선
contour = plt.contourf(
    W,
    B,
    Loss,
    levels=30,
    cmap="viridis"
)

plt.colorbar(contour, label="Loss")

# gradient descent 경로
plt.plot(
    w_history,
    b_history,
    color="red",
    linewidth=2,
    label="gradient descent path"
)

# 시작점
plt.scatter(
    w_history[0],
    b_history[0],
    color="white",
    edgecolor="black",
    s=100,
    label="start"
)

# 마지막점
plt.scatter(
    w_history[-1],
    b_history[-1],
    color="red",
    edgecolor="black",
    s=100,
    label="end"
)

plt.xlabel("w")
plt.ylabel("b")
plt.title("Gradient Descent on Loss Surface")
plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D


# -----------------------
# 데이터
# -----------------------

x = np.array([1., 2., 3., 4.])
y = np.array([3., 5., 7., 9.])


# -----------------------
# 초기 parameter
# -----------------------

w = 0.0
b = 0.0


# -----------------------
# hyperparameter
# -----------------------

learning_rate = 0.01
epochs = 100


n = len(x)


# -----------------------
# Gradient Descent
# -----------------------

w_history = []
b_history = []
loss_history = []


for epoch in range(epochs):

    # prediction
    y_pred = w * x + b

    # loss
    loss = np.mean((y_pred - y) ** 2)

    # 기록
    w_history.append(w)
    b_history.append(b)
    loss_history.append(loss)

    # gradient
    dw = (2 / n) * np.sum((y_pred - y) * x)
    db = (2 / n) * np.sum(y_pred - y)

    # update
    w = w - learning_rate * dw
    b = b - learning_rate * db


# -----------------------
# Loss Surface 만들기
# -----------------------

w_values = np.linspace(-1, 4, 100)
b_values = np.linspace(-2, 4, 100)

W, B = np.meshgrid(w_values, b_values)

Loss = np.zeros_like(W)


for i in range(W.shape[0]):
    for j in range(W.shape[1]):

        y_pred = W[i, j] * x + B[i, j]

        Loss[i, j] = np.mean(
            (y_pred - y) ** 2
        )


# -----------------------
# 3D 그래프
# -----------------------

fig = plt.figure(figsize=(10, 7))

ax = fig.add_subplot(
    111,
    projection="3d"
)


# Loss surface
ax.plot_surface(
    W,
    B,
    Loss,
    cmap="viridis",
    alpha=0.7
)


# Gradient Descent 경로
ax.plot(
    w_history,
    b_history,
    loss_history,
    color="red",
    marker="o",
    markersize=3,
    linewidth=2,
    label="Gradient Descent"
)


# 시작점
ax.scatter(
    w_history[0],
    b_history[0],
    loss_history[0],
    color="blue",
    s=80,
    label="Start"
)


# 최종점
ax.scatter(
    w_history[-1],
    b_history[-1],
    loss_history[-1],
    color="red",
    s=80,
    label="End"
)


# 축
ax.set_xlabel("w")
ax.set_ylabel("b")
ax.set_zlabel("Loss")

ax.set_title(
    "Loss Surface and Gradient Descent"
)

ax.legend()

plt.show()
```

```python
def softmax(logits):
    logits = logits - np.max(logits)

    exp_values = np.exp(logits)

    return exp_values / np.sum(exp_values)


def cross_entropy(probabilities, target):
    return -np.log(probabilities[target])


logits = np.array([2.0, 4.0, 1.0])

target = 1

probabilities = softmax(logits)

loss = cross_entropy(probabilities, target)

print("probabilities:", probabilities)
print("loss:", loss)
```

```python
def step(z):
    return 1 if z >= 0 else 0


def and_gate(x1, x2):
    w1 = 1.0
    w2 = 1.0
    b = -1.5

    z = w1 * x1 + w2 * x2 + b

    return step(z)


for x1, x2 in [
    (0, 0),
    (0, 1),
    (1, 0),
    (1, 1)
]:
    print(x1, x2, "->", and_gate(x1, x2))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# -------------------------
# 가중치와 편향
# -------------------------

w1 = 2
w2 = 3
w3 = 4
b = 1


# -------------------------
# x1, x2 범위
# -------------------------

x1 = np.linspace(-5, 5, 50)
x2 = np.linspace(-5, 5, 50)

X1, X2 = np.meshgrid(x1, x2)


# -------------------------
# 그래프 함수
# -------------------------

def draw_plane(x3):

    # z = w1*x1 + w2*x2 + w3*x3 + b
    Z = w1 * X1 + w2 * X2 + w3 * x3 + b

    # 그래프 생성
    fig = plt.figure(figsize=(9, 7))
    ax = fig.add_subplot(111, projection='3d')

    # 평면 그리기
    ax.plot_surface(
        X1,
        X2,
        Z,
        cmap='viridis',
        alpha=0.8
    )

    # 축 이름
    ax.set_xlabel('x1')
    ax.set_ylabel('x2')
    ax.set_zlabel('z')

    # 제목
    ax.set_title(
        f'z = {w1}x1 + {w2}x2 + {w3}x3 + {b}\n'
        f'x3 = {x3:.1f}'
    )

    plt.show()


# -------------------------
# x3 슬라이더
# -------------------------

interact(
    draw_plane,
    x3=FloatSlider(
        min=-5,
        max=5,
        step=0.1,
        value=0,
        description='x3'
    )
);
```

```python
import numpy as np


def softmax(x):
    exp_x = np.exp(
        x - np.max(x, axis=-1, keepdims=True)
    )

    return exp_x / np.sum(
        exp_x,
        axis=-1,
        keepdims=True
    )


def layer_norm(x, eps=1e-5):
    mean = np.mean(
        x,
        axis=-1,
        keepdims=True
    )

    var = np.mean(
        (x - mean) ** 2,
        axis=-1,
        keepdims=True
    )

    return (
        x - mean
    ) / np.sqrt(
        var + eps
    )
```

```python
def self_attention(
    X,
    W_Q,
    W_K,
    W_V,
    W_O,
    num_heads,
    causal=True
):
    B, S, D = X.shape

    head_dim = D // num_heads

    # 1. Q, K, V
    Q = W_Q(X)
    K = W_K(X)
    V = W_V(X)

    # 2. Split heads
    Q = Q.view(B, S, num_heads, head_dim)
    K = K.view(B, S, num_heads, head_dim)
    V = V.view(B, S, num_heads, head_dim)

    # 3. (B, H, S, Dh)
    Q = Q.transpose(1, 2)
    K = K.transpose(1, 2)
    V = V.transpose(1, 2)

    # 4. QK^T
    scores = Q @ K.transpose(-2, -1)

    # 5. Scale
    scores = scores / math.sqrt(head_dim)

    # 6. Causal mask
    if causal:
        mask = torch.triu(
            torch.ones(
                S,
                S,
                device=X.device,
                dtype=torch.bool
            ),
            diagonal=1
        )

        scores = scores.masked_fill(
            mask,
            float("-inf")
        )

    # 7. Softmax
    weights = torch.softmax(
        scores,
        dim=-1
    )

    # 8. Weighted sum of V
    context = weights @ V

    # 9. Merge heads
    context = context.transpose(1, 2)

    context = context.contiguous().view(
        B,
        S,
        D
    )

    # 10. Output projection
    output = W_O(context)

    return output, weights
```

```python
import torch
import torch.nn as nn
import math


class SimpleTransformerBlock(nn.Module):

    def __init__(
        self,
        d_model=8,
        num_heads=2,
        d_ff=32
    ):
        super().__init__()

        assert d_model % num_heads == 0

        self.d_model = d_model
        self.num_heads = num_heads
        self.head_dim = (
            d_model // num_heads
        )

        self.W_Q = nn.Linear(
            d_model,
            d_model,
            bias=False
        )

        self.W_K = nn.Linear(
            d_model,
            d_model,
            bias=False
        )

        self.W_V = nn.Linear(
            d_model,
            d_model,
            bias=False
        )

        self.W_O = nn.Linear(
            d_model,
            d_model,
            bias=False
        )

        self.norm1 = nn.LayerNorm(
            d_model
        )

        self.norm2 = nn.LayerNorm(
            d_model
        )

        self.ffn = nn.Sequential(
            nn.Linear(
                d_model,
                d_ff
            ),
            nn.ReLU(),
            nn.Linear(
                d_ff,
                d_model
            )
        )

    def attention(self, X):

        B, S, D = X.shape

        Q = self.W_Q(X)
        K = self.W_K(X)
        V = self.W_V(X)

        Q = Q.view(
            B,
            S,
            self.num_heads,
            self.head_dim
        )

        K = K.view(
            B,
            S,
            self.num_heads,
            self.head_dim
        )

        V = V.view(
            B,
            S,
            self.num_heads,
            self.head_dim
        )

        Q = Q.transpose(1, 2)
        K = K.transpose(1, 2)
        V = V.transpose(1, 2)

        scores = (
            Q @ K.transpose(-2, -1)
        )

        scores = (
            scores /
            math.sqrt(self.head_dim)
        )

        mask = torch.triu(
            torch.ones(
                S,
                S,
                device=X.device,
                dtype=torch.bool
            ),
            diagonal=1
        )

        scores = scores.masked_fill(
            mask,
            float("-inf")
        )

        weights = torch.softmax(
            scores,
            dim=-1
        )

        context = weights @ V

        context = context.transpose(
            1,
            2
        )

        context = (
            context
            .contiguous()
            .view(B, S, D)
        )

        output = self.W_O(
            context
        )

        return output, weights

    def forward(self, X):

        attention_output, weights = (
            self.attention(X)
        )

        X = self.norm1(
            X + attention_output
        )

        ffn_output = self.ffn(X)

        X = self.norm2(
            X + ffn_output
        )

        return X, weights
```
