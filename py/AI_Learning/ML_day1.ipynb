```python
import numpy as np
import matplotlib.pyplot as plt

# 데이터
x = np.array([1., 2., 3., 4.])
y = np.array([3., 5., 7., 9.])

# 초기 parameter
w = 0.0
b = 0.0

# hyperparameter
learning_rate = 0.01
epochs = 1000

n = len(x)

# 기록용
w_history = []
b_history = []
loss_history = []

for epoch in range(epochs):

    # 1. prediction
    y_pred = w * x + b

    # 2. loss
    loss = np.mean((y_pred - y) ** 2)

    # 기록
    w_history.append(w)
    b_history.append(b)
    loss_history.append(loss)

    # 3. gradient
    dw = (2 / n) * np.sum((y_pred - y) * x)
    db = (2 / n) * np.sum(y_pred - y)

    # 4. parameter update
    w = w - learning_rate * dw
    b = b - learning_rate * db


# -----------------------
# 그래프
# -----------------------

plt.figure(figsize=(8, 5))

plt.plot(w_history, loss_history, color="blue", alpha=0.7)

# 각 epoch의 점
plt.scatter(
    w_history[::20],
    loss_history[::20],
    color="red",
    s=20
)

# 시작점
plt.scatter(
    w_history[0],
    loss_history[0],
    color="green",
    s=100,
    label="start"
)

# 마지막점
plt.scatter(
    w_history[-1],
    loss_history[-1],
    color="black",
    s=100,
    label="end"
)

plt.xlabel("w")
plt.ylabel("loss")
plt.title("Gradient Descent: w vs loss")
plt.legend()
plt.grid()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 데이터
x = np.array([1., 2., 3., 4.])
y = np.array([3., 5., 7., 9.])

# 초기 parameter
w = 0.0
b = 0.0

learning_rate = 0.01
epochs = 1000

n = len(x)

# 기록
w_history = []
b_history = []
loss_history = []

for epoch in range(epochs):

    # prediction
    y_pred = w * x + b

    # loss
    loss = np.mean((y_pred - y) ** 2)

    # 기록
    w_history.append(w)
    b_history.append(b)
    loss_history.append(loss)

    # gradient
    dw = (2 / n) * np.sum((y_pred - y) * x)
    db = (2 / n) * np.sum(y_pred - y)

    # update
    w = w - learning_rate * dw
    b = b - learning_rate * db


# -----------------------
# loss surface 계산
# -----------------------

w_values = np.linspace(-1, 4, 100)
b_values = np.linspace(-2, 4, 100)

W, B = np.meshgrid(w_values, b_values)

Loss = np.zeros_like(W)

for i in range(W.shape[0]):
    for j in range(W.shape[1]):
        y_pred = W[i, j] * x + B[i, j]
        Loss[i, j] = np.mean((y_pred - y) ** 2)


# -----------------------
# 그래프
# -----------------------

plt.figure(figsize=(8, 6))

# loss 등고선
contour = plt.contourf(
    W,
    B,
    Loss,
    levels=30,
    cmap="viridis"
)

plt.colorbar(contour, label="Loss")

# gradient descent 경로
plt.plot(
    w_history,
    b_history,
    color="red",
    linewidth=2,
    label="gradient descent path"
)

# 시작점
plt.scatter(
    w_history[0],
    b_history[0],
    color="white",
    edgecolor="black",
    s=100,
    label="start"
)

# 마지막점
plt.scatter(
    w_history[-1],
    b_history[-1],
    color="red",
    edgecolor="black",
    s=100,
    label="end"
)

plt.xlabel("w")
plt.ylabel("b")
plt.title("Gradient Descent on Loss Surface")
plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import clear_output, display
import time

# 데이터
x = np.array([1., 2., 3., 4.])
y = np.array([3., 5., 7., 9.])

# 초기 parameter
w = 0.0
b = 0.0

# hyperparameter
learning_rate = 0.01
epochs = 50

n = len(x)

# 기록
w_history = []
loss_history = []


for epoch in range(epochs):

    # 1. prediction
    y_pred = w * x + b

    # 2. loss
    loss = np.mean((y_pred - y) ** 2)

    # 기록
    w_history.append(w)
    loss_history.append(loss)

    # -----------------------
    # 그래프 그리기
    # -----------------------

    clear_output(wait=True)

    fig, ax = plt.subplots(figsize=(8, 5))

    # 지금까지 이동한 경로
    ax.plot(
        w_history,
        loss_history,
        color="blue"
    )

    # 현재 위치
    ax.scatter(
        w,
        loss,
        color="red",
        s=80
    )

    ax.set_xlim(-0.2, 2.2)
    ax.set_ylim(0, 45)

    ax.set_xlabel("w")
    ax.set_ylabel("loss")

    ax.set_title(
        f"epoch={epoch}, "
        f"w={w:.4f}, "
        f"b={b:.4f}, "
        f"loss={loss:.6f}"
    )

    ax.grid()

    display(fig)
    plt.close(fig)

    # 잠깐 멈춤
    time.sleep(0.1)

    # -----------------------
    # 3. gradient
    # -----------------------

    dw = (2 / n) * np.sum((y_pred - y) * x)
    db = (2 / n) * np.sum(y_pred - y)

    # -----------------------
    # 4. parameter update
    # -----------------------

    w = w - learning_rate * dw
    b = b - learning_rate * db
```

```python
import numpy as np
import matplotlib.pyplot as plt

from mpl_toolkits.mplot3d import Axes3D


# -----------------------
# 데이터
# -----------------------

x = np.array([1., 2., 3., 4.])
y = np.array([3., 5., 7., 9.])


# -----------------------
# 초기 parameter
# -----------------------

w = 0.0
b = 0.0


# -----------------------
# hyperparameter
# -----------------------

learning_rate = 0.01
epochs = 100


n = len(x)


# -----------------------
# Gradient Descent
# -----------------------

w_history = []
b_history = []
loss_history = []


for epoch in range(epochs):

    # prediction
    y_pred = w * x + b

    # loss
    loss = np.mean((y_pred - y) ** 2)

    # 기록
    w_history.append(w)
    b_history.append(b)
    loss_history.append(loss)

    # gradient
    dw = (2 / n) * np.sum((y_pred - y) * x)
    db = (2 / n) * np.sum(y_pred - y)

    # update
    w = w - learning_rate * dw
    b = b - learning_rate * db


# -----------------------
# Loss Surface 만들기
# -----------------------

w_values = np.linspace(-1, 4, 100)
b_values = np.linspace(-2, 4, 100)

W, B = np.meshgrid(w_values, b_values)

Loss = np.zeros_like(W)


for i in range(W.shape[0]):
    for j in range(W.shape[1]):

        y_pred = W[i, j] * x + B[i, j]

        Loss[i, j] = np.mean(
            (y_pred - y) ** 2
        )


# -----------------------
# 3D 그래프
# -----------------------

fig = plt.figure(figsize=(10, 7))

ax = fig.add_subplot(
    111,
    projection="3d"
)


# Loss surface
ax.plot_surface(
    W,
    B,
    Loss,
    cmap="viridis",
    alpha=0.7
)


# Gradient Descent 경로
ax.plot(
    w_history,
    b_history,
    loss_history,
    color="red",
    marker="o",
    markersize=3,
    linewidth=2,
    label="Gradient Descent"
)


# 시작점
ax.scatter(
    w_history[0],
    b_history[0],
    loss_history[0],
    color="blue",
    s=80,
    label="Start"
)


# 최종점
ax.scatter(
    w_history[-1],
    b_history[-1],
    loss_history[-1],
    color="red",
    s=80,
    label="End"
)


# 축
ax.set_xlabel("w")
ax.set_ylabel("b")
ax.set_zlabel("Loss")

ax.set_title(
    "Loss Surface and Gradient Descent"
)

ax.legend()

plt.show()
```

```python
def softmax(logits):
    logits = logits - np.max(logits)

    exp_values = np.exp(logits)

    return exp_values / np.sum(exp_values)


def cross_entropy(probabilities, target):
    return -np.log(probabilities[target])


logits = np.array([2.0, 4.0, 1.0])

target = 1

probabilities = softmax(logits)

loss = cross_entropy(probabilities, target)

print("probabilities:", probabilities)
print("loss:", loss)
```
