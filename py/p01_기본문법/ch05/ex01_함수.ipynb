```python
def sum(*args):
    result=0
    print(args)
    print(type(args))
    num = [x for x in args if isinstance(x, (int, float))]
    print(num)
    print(type(num))
    for n in num:
        result += n

    return result, num
```

```python
sum(1,2,3,'a',5, 10, None, 20, 'b', 30, 'c')
```

```python
sum(10,40)
```

```python
def c_to_f(celsius):
    f = (celsius * 9/5)+32
    print(f'섭씨 {celsius}도는 화씨 {f}도 입니다.')

    return f
```

```python
c_to_f(30)
```
