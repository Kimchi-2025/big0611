```python
plt.figure(figsize=(12,20))

plt.subplot(5,1,1)
plt.title('Total bill (hist)')
sns.histplot(df, x='total_bill')

plt.subplot(5,1,2)
plt.title('Tip (hist)')
sns.histplot(df, x='tip')

plt.subplot(5,1,3)
plt.title('Sex (count)')
sns.countplot(df, x='sex')

plt.subplot(5,1,4)
plt.title('Smoker (count)')
sns.countplot(df, x='smoker')

plt.subplot(5,1,5)
plt.title('Smoker (count)')
sns.barplot(df, x='smoker', y='total_bill', estimator=np.median)

plt.subplots_adjust(hspace=0.5)
```
