```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x = list(range(1,13))
y = list(range(10,130,10))
```

```python
plt.bar(x,y)

plt.title('Monthly expenses')
plt.xlabel('Month')
plt.ylabel('Cost($)')
plt.legend(['cost'])
plt.grid(True, axis='y')
plt.show()
```

```python
plt.figure(figsize=(12,10))

plt.subplot(3,2,1)
plt.bar(x,y)
plt.title('Monthly expenses')
plt.xlabel('Month')
plt.ylabel('Cost($)')
plt.legend(['cost'])
plt.grid(True, axis='y')

plt.subplot(3,2,2)
plt.bar(x,y)
plt.title('Monthly expenses')
plt.xlabel('Month')
plt.ylabel('Cost($)')
plt.legend(['cost'])
plt.grid(True, axis='y')

plt.subplot(3,2,3)
plt.bar(x,y)
plt.title('Monthly expenses')
plt.xlabel('Month')
plt.ylabel('Cost($)')
plt.legend(['cost'])
plt.grid(True, axis='y')

plt.subplot(3,2,4)
plt.bar(x,y)
plt.title('Monthly expenses')
plt.xlabel('Month')
plt.ylabel('Cost($)')
plt.legend(['cost'])
plt.grid(True, axis='y')

plt.subplot(3,2,5)
plt.bar(x,y)
plt.title('월별 지출')
plt.xlabel('Month')
plt.ylabel('Cost($)')
plt.legend(['cost'])
plt.grid(True, axis='y')

plt.tight_layout()
plt.show()
```
