```python
rng = np.random.default_rng()
```

```python
data = rng.normal(0,1, 10000)
```

```python
data.mean(), data.std()
```

```python
sns.histplot(data, kde=True)
```

```python
from scipy.stats import norm
```

```python
mu = data.mean()
sigma = data.std()
mu, sigma
```

```python
data = np.sort(data)
data
```

```python
x = data
y = norm.pdf(data, mu, sigma)

plt.plot(x, y)
```

```python
x = data
y = norm.pdf(data, 0, 1)

sns.histplot(data, bins=30, stat='density')
plt.plot(x, y, color='red')
plt.show()
```
