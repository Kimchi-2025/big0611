```python
arr = df['Price_up'].values
```

```python
max_len = 0       # 가장 긴 길이
max_start = -1    # 가장 긴 구간의 시작 인덱스

current_len = 0
current_start = 0

for i, x in enumerate(arr):
    if x == 1:
        if current_len == 0:
            current_start = i      # 새로운 연속 구간 시작
        current_len += 1

        if current_len > max_len:
            max_len = current_len
            max_start = current_start
    else:
        current_len = 0

print(f'최대 연속 개수 : {max_len}')
print(f'시작 인덱스   : {max_start}')
print(f'끝 인덱스     : {max_start + max_len - 1}')
```
