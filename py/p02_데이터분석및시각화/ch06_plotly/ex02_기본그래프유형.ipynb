```python
import plotly.express as px
import pandas as pd
import numpy as np

# 시계열 데이터 생성
dates = pd.date_range('2024-01-01', periods=30, freq='D')
values = np.cumsum(np.random.randn(30)) + 100

time_series = pd.DataFrame({
    '날짜': dates,
    '주가': values
})

fig = px.line(time_series, x='날짜', y='주가',
              title='일별 주가 변동',
              markers=True)
fig.show()
```
