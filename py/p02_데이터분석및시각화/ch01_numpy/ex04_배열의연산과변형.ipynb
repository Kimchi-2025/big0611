```python
arr = np.array([1,2,3,4])
m = arr.mean()
n = len(arr)

var = (np.square(arr - m).sum())/n
std = np.sqrt((np.square(arr - m).sum())) / np.sqrt(n)
print("var: ", var)
print("std: ", std)
print('se: ', std/np.sqrt(n))
```

```python
price = np.array([100,102,98,105,103,99,101])

def moving_average(data, window):
    return np.convolve(data, np.ones(window)/window, mode='full')

ma_3 = moving_average(price,3)
print(ma_3)
```
