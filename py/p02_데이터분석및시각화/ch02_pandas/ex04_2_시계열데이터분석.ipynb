```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

flights = sns.load_dataset("flights")

passengers = flights["passengers"]

# 로그 변환 + 일반 1차 차분 + 12개월 계절 차분
stationary = (
    np.log(passengers)
    .diff(1)
    .diff(12)
    .dropna()
)

plot_acf(stationary, lags=40)
plt.show()

plot_pacf(stationary, lags=40, method="ywm")
plt.show()
```

```python
from pmdarima import auto_arima

model = auto_arima(
    ts,
    seasonal=True,
    m=12,          # 월별 데이터
    start_p=0,
    start_q=0,
    max_p=3,
    max_q=3,
    max_P=2,
    max_Q=2,
    d=1,
    D=1,
    trace=True,    # 후보 모델 출력
    stepwise=True,
    information_criterion='aic'
)

print(model.summary())
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

model = SARIMAX(
    ts,
    order=(0,1,1),
    seasonal_order=(2,1,0,12)
)

fit = model.fit()

print(fit.summary())
```
