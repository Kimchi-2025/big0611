```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
raw = sns.load_dataset('flights')
raw
```

```python
df = raw.copy()
df
```

```python
df.info()
```

```python
df.head()
```

```python
df['date'] = pd.to_datetime(df['year'].astype(str) + '-' + df['month'].astype(str))
```

```python
df
```

```python
df.set_index('date', inplace=True)
```

```python
df
```

```python
plt.figure(figsize=(12,5))
df['passengers'].plot(label='base')
df['passengers'].rolling(6).mean().plot(label='6-mon Ave')
df['passengers'].rolling(12).mean().plot(label='12-mon Ave')
plt.legend()
plt.title('Flights')
```

```python
from statsmodels.graphics.tsaplots import plot_acf
```

```python
passengers = df["passengers"]
```

```python
plt.figure(figsize=(12,5))

plot_acf(
    passengers,
    lags=40
)

plt.show()
```

```python
diff1 = passengers.diff().dropna()

plt.figure(figsize=(12,5))

plot_acf(
    diff1,
    lags=40
)

plt.show()
```

```python
from statsmodels.graphics.tsaplots import plot_pacf
```

```python
plt.figure(figsize=(12,5))

plot_pacf(
    diff1,
    lags=40
)

plt.show()
```
