```python
def make_group(row):
    if (row['survived'] == 1) and (row['sex'] == 'female') and (row['pclass'] in [1,2]):
        return 'A'

    elif (row['survived'] == 1) and (row['sex'] == 'male') and (row['age'] < 20):
        return 'B'
    
    elif (row['survived'] == 0) and (row['sex'] == 'male') and (row['pclass'] == 3):
        return 'C'

    else:
        return 'D'
```

```python
def make_group2(row):
    if (row['sex'] == 'male') and (row['class'] == 'Third') and (row['alone']==True):
        return 'High Risk'

    elif (row['sex'] == 'female') and (row['class'] == 'Third'):
        return 'Medium Risk'

    elif (row['sex'] == 'female') and (row['class'] == 'First'):
        return 'Safe'

    else:
        return 'Others'
```

```python
def make_group3(row):
    if (row['class'] == 'First') and (row['sex'] == 'female') and (row['fare'] >= 80) and (row['alive'] == 'yes'):
        return 'VIP'

    elif (row['alive'] == 'yes') and (row['age'] < 18) and (row['alone'] == False):
        return 'Young Survivor'
    
    elif (row['sex'] == 'male') and (row['class'] == 'Third') and (row['alone'] == True) and (row['age'] >= 18):
        return 'High Risk'

    elif (row['sibsp'] + row['parch'] >= 2):
        return 'Family Passenger'

    else:
        return 'Others'
```
