```python
conditions = [
    data['total_sales'] >= 10000000,
    data['total_sales'] >= 5000000,
    data['total_sales'] >= 2000000
]

choices = ['대형거래', '중형거래', '소형거래']
```

```python
data['size'] = np.select(conditions, choices, default='미니거래')
```

```python
data
```

```python
def transaction(row):
    if row.total_sales >= 10000000:
        return '대형거래'
    elif row.total_sales >= 5000000:
        return '중형거래'
    elif row.total_sales >= 2000000:
        return '소형거래'
    else:
        return '미니거래'
```

```python
data['trans'] = data.apply(transaction, axis=1)
```
