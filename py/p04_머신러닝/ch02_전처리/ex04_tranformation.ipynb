```python
# 데이터 변환
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.preprocessing import LabelEncoder, MinMaxScaler, StandardScaler

data = {
    '이름': ['오리', '오리', '미금', '기흥', '미금', '기흥'],
    '과목': ['수학', '과학', '수학', '과학', '과학', '수학'],
    '점수': [85, 90, 78, 88, 95, 80]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.groupby('과목')['점수'].mean()
```

```python
pd.pivot(df, index='이름', columns='과목', values='점수')
```

```python
pd.pivot_table(df, index='이름', columns='과목', values='점수', aggfunc='mean')
```

```python
pd.crosstab(df['이름'], df['과목'], values=df['점수'], aggfunc='sum')
```

```python
# 2. 데이터 인코딩: 레이블 인코딩, one-hot 인코딩
# 예시 데이터
data = ['Red', 'Green', 'Blue', 'Green', 'Red']

# LabelEncoder 객체 생성
encoder = LabelEncoder()

# 'data'를 숫자형으로 변환
encoded_data = encoder.fit_transform(data)
print(encoded_data)
```

```python
data = {
    'Gender': ['Male', 'Female', 'Female', 'Male'],
    'City': ['Seoul', 'Busan', 'Incheon', 'Seoul']
}

df = pd.DataFrame(data)
print(df)
```

```python
# 범주형 변수가 포함된 열에 대해 더미 변수를 생성
df_dummies = pd.get_dummies(df)
print(df_dummies)
```

```python
pd.get_dummies(df).astype(int)
```

```python
# 특정 열(Gender)에 대해서만 더미 변수를 생성
df_dummies_gender = pd.get_dummies(df, columns=['Gender'])
print(df_dummies_gender)
```

```python
data = {
    '나이': [25, 30, 45, 50, 60],
    '연봉': [3500, 4000, 5000, 5500, 6500]
}

df = pd.DataFrame(data)
print(df)
```

```python
min_max_scaler = MinMaxScaler()
df_minmax = pd.DataFrame(min_max_scaler.fit_transform(df), columns=df.columns)
print('Min-Max 정규화된 데이터')
df_minmax
```

```python
# 표준화 수행
standard_scaler = StandardScaler()
df_standard = pd.DataFrame(standard_scaler.fit_transform(df), columns=df.columns)
print('표준화된 데이터')
df_standard
```

```python
# 사이킷런 패키지는 다양한 데이터셋을 제공한다.
# 붓꽃 데이터셋(iris data)
# - 꽃받침 길이와 너비, 꽃잎 길이와 너비의 4개의 특징 데이터에 따른 품종을 구분한 데이터로 총 150개의 값이 존재한다.
# 데이터 불러오기
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
print(df)
df['species'] = iris.target
```

```python
df.describe()
```

```python
X = df.drop(columns='species')
```

```python
X
```

```python
mm_scaler = MinMaxScaler()
mm_data = mm_scaler.fit_transform(X)
```
