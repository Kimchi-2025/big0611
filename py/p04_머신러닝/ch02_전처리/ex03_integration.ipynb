```python
import numpy as np
import pandas as pd
```

```python
# 샘플 데이터 생성
df_user = pd.DataFrame({'id': [1, 2, 3], 'name': ['Alice', 'Bob', 'Charlie']})
df_order = pd.DataFrame({'id': [1, 2, 4], 'product': ['Notebook', 'Mouse', 'Monitor']})

# 'id' 컬럼을 기준으로 왼쪽(df_user) 데이터 기준 병합 (가장 많이 씀!)
merged_df = pd.merge(df_user, df_order, on='id', how='left')
print(merged_df)
```

```python
# 샘플 데이터 생성 (1월 매출과 2월 매출 데이터)
df_jan = pd.DataFrame({'sales': [100, 150], 'profit': [10, 15]}, index=['Prod_A', 'Prod_B'])
df_feb = pd.DataFrame({'sales': [120, 170], 'profit': [12, 18]}, index=['Prod_A', 'Prod_B'])

# 위아래로 데이터 이어붙이기
total_sales = pd.concat([df_jan, df_feb], axis=0)
print(total_sales)
```
