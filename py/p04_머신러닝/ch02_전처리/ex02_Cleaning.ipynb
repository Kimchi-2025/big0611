### 결측치 처리
결측치?  
데이터 수집 문제 또는 데이터 입력 오류와 같은 다양한 이유로 관측되어야 할 값을 얻지 못한 데이터

```python
import pandas as pd
import numpy as np

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, np.nan, 35, 30],         # Bob의 나이 결측치
    'Salary': [5000, 4000, np.nan, 6000] # Charlie의 급여 결측치
}
df = pd.DataFrame(data)
```
