```python
scores = [80, 90, 70, 65, 95, 100, 90, 80, 75, 80]
new_scores = []
index = 0

while index < len(scores):
    if scores[index] < 100:
        new = scores[index] +5
    else:
        new = scores[index]
    
    new_scores.append(new)
    index += 1

print(scores)
print(new_scores)
```

```python
inventory = {"사과":50, "바나나":30, "오렌지":25}
inventory["포도"] = 40
print('포도 재고 추가 완료')

print("==현재 재고 현황==")
for key, value in inventory.items():
    print(f"{key}: {value}")

product = '사과'
if product in inventory:
    print(f'{product} 현재 재고 {inventory[product]}개')

print("==재고 부족 상품==")
for key, value in inventory.items():
    if value < 30:
        print(f'{key}: {value}개(재고부족)')

product = '사과'
sales = 10
print(f'{product} {sales}개 판매 완료')
inventory[product] -= sales
print(f'{product} 현재 재고: {inventory[product]}개')
```
